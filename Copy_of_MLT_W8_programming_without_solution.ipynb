# Bernoulli naive Bayes

Run the below cell to get the following variables:

`X` = Data matrix of shape $(n, d)$. All the features are binary taking values $0$ or $1$.

`y` = label vector. Labels are $0$ and $1$.

```python
rng = np.random.default_rng(seed=1)
X1 = np.concatenate((rng.binomial(size = 50,n = 1, p =0.7), rng.binomial(size = 50,n = 1, p =0.2))).reshape(-1, 1)
X2 = np.concatenate((rng.binomial(size = 50,n = 1, p =0.6), rng.binomial(size = 50,n = 1, p =0.1))).reshape(-1, 1)
X3 = np.concatenate((rng.binomial(size = 50,n = 1, p =0.6), rng.binomial(size = 50,n = 1, p =0.2))).reshape(-1, 1)
X4 = np.concatenate((rng.binomial(size = 50,n = 1, p =0.8), rng.binomial(size = 50,n = 1, p =0.1))).reshape(-1, 1)

X = np.column_stack((X1,X2,X3,X4))

y = np.concatenate((np.zeros(50, dtype= int), np.ones(50, dtype = int))).reshape(-1, 1)
permute = rng.permuted(range(100))

X = X[permute]
y = y[permute]
```

## Question 1
If we train the naive Bayes model on the dataset, What will be the value of $\hat{p}$, the estimate for $P(Y=1)$?



```python
# Question 1
# Estimate for P(Y=1) is the proportion of samples where y=1
p_y1 = np.sum(y == 1) / len(y)
print(f"The estimate for P(Y=1) is: {p_y1}")
```

## Question 2
What will be the value of $\hat{p}_0^0$, the estimate of $P(f_0=1|y=0)$?  Write your answer correct to two decimal places.



```python
# Question 2
# Estimate of P(f_0=1|y=0)
# Filter X where y=0
X_y0 = X[y.flatten() == 0]
# Count where f_0 (first feature) is 1 for y=0
p_f0_y0 = np.sum(X_y0[:, 0] == 1) / len(X_y0)
print(f"The estimate for P(f_0=1|y=0) is: {p_f0_y0:.2f}")
```

## Question 3
What will be the value of $\hat{p}_0^1$, the estimate of $P(f_0=1|y=1)$?  Write your answer correct to two decimal places.



```python
# Question 3
# Estimate of P(f_0=1|y=1)
# Filter X where y=1
X_y1 = X[y.flatten() == 1]
# Count where f_0 (first feature) is 1 for y=1
p_f0_y1 = np.sum(X_y1[:, 0] == 1) / len(X_y1)
print(f"The estimate for P(f_0=1|y=1) is: {p_f0_y1:.2f}")
```

## Question 4

## Question 5

What will be the predicted label for the point $[1, 0, 1, 0]$?



```python
# Question 5
# Input point
x_q5 = np.array([1, 0, 1, 0])

# Prior probabilities
p_y0 = 1 - p_y1 # from Question 1

# Conditional probabilities P(f_j=1|y=0) and P(f_j=1|y=1) for all features
# Re-filter X_y0 and X_y1 to ensure they are consistent with the current X and y (though they should be)
X_y0 = X[y.flatten() == 0]
X_y1 = X[y.flatten() == 1]

# Calculate all P(f_j=1|y=0)
p_f1_y0_all = np.array([np.sum(X_y0[:, j] == 1) / len(X_y0) for j in range(X.shape[1])])

# Calculate all P(f_j=1|y=1)
p_f1_y1_all = np.array([np.sum(X_y1[:, j] == 1) / len(X_y1) for j in range(X.shape[1])])

# Calculate P(x|y=0) - likelihood of x given y=0
likelihood_y0 = 1.0
for j in range(X.shape[1]):
    if x_q5[j] == 1:
        likelihood_y0 *= p_f1_y0_all[j]
    else: # x_q5[j] == 0
        likelihood_y0 *= (1 - p_f1_y0_all[j])

# Calculate P(x|y=1) - likelihood of x given y=1
likelihood_y1 = 1.0
for j in range(X.shape[1]):
    if x_q5[j] == 1:
        likelihood_y1 *= p_f1_y1_all[j]
    else: # x_q5[j] == 0
        likelihood_y1 *= (1 - p_f1_y1_all[j])

# Calculate posterior probabilities (proportional)
posterior_y0 = likelihood_y0 * p_y0
posterior_y1 = likelihood_y1 * p_y1

# Predict label
predicted_label_q5 = 0 if posterior_y0 > posterior_y1 else 1

print(f"Point: {x_q5}")
print(f"P(Y=0): {p_y0:.2f}")
print(f"P(Y=1): {p_y1:.2f}")
print(f"P(f_j=1|y=0) for j=0..3: {p_f1_y0_all}")
print(f"P(f_j=1|y=1) for j=0..3: {p_f1_y1_all}")
print(f"P(x|y=0): {likelihood_y0}")
print(f"P(x|y=1): {likelihood_y1}")
print(f"P(y=0|x) proportional to: {posterior_y0}")
print(f"P(y=1|x) proportional to: {posterior_y1}")
print(f"Predicted label for [1, 0, 1, 0]: {predicted_label_q5}")
```

```python
# Question 6
# Input point
x_q6 = np.array([1, 0, 1, 1])

# Prior probabilities p_y0, p_y1 are already calculated from Question 1 (p_y1) and derived (p_y0).
# Conditional probabilities p_f1_y0_all and p_f1_y1_all are already calculated in Question 5.

# Calculate P(x|y=0) - likelihood of x given y=0 for x_q6
likelihood_y0_q6 = 1.0
for j in range(X.shape[1]):
    if x_q6[j] == 1:
        likelihood_y0_q6 *= p_f1_y0_all[j]
    else: # x_q6[j] == 0
        likelihood_y0_q6 *= (1 - p_f1_y0_all[j])

# Calculate P(x|y=1) - likelihood of x given y=1 for x_q6
likelihood_y1_q6 = 1.0
for j in range(X.shape[1]):
    if x_q6[j] == 1:
        likelihood_y1_q6 *= p_f1_y1_all[j]
    else: # x_q6[j] == 0
        likelihood_y1_q6 *= (1 - p_f1_y1_all[j])

# Calculate posterior probabilities (proportional)
posterior_y0_q6 = likelihood_y0_q6 * p_y0
posterior_y1_q6 = likelihood_y1_q6 * p_y1

# Predict label
predicted_label_q6 = 0 if posterior_y0_q6 > posterior_y1_q6 else 1

print(f"Point: {x_q6}")
print(f"P(Y=0): {p_y0:.2f}")
print(f"P(Y=1): {p_y1:.2f}")
print(f"P(f_j=1|y=0) for j=0..3: {p_f1_y0_all}")
print(f"P(f_j=1|y=1) for j=0..3: {p_f1_y1_all}")
print(f"P(x|y=0): {likelihood_y0_q6}")
print(f"P(x|y=1): {likelihood_y1_q6}")
print(f"P(y=0|x) proportional to: {posterior_y0_q6}")
print(f"P(y=1|x) proportional to: {posterior_y1_q6}")
print(f"Predicted label for [1, 0, 1, 1]: {predicted_label_q6}")
```

# Gaussian naive Bayes

Run the below cell to get the following variables:

`X_train` = Training dataset of the shape $(n, d)$. All the examples are coming from multivariate gaussian distribution.

`y_train` = label vector for corresponding training examples. labels are $0$ and $1$.

`X_test` = Test dataset of the shape $(m, d)$, where $m$ is the number of examples in the test dataset. All the examples are coming from multivariate gaussian distribution.

`y_test` = label vector for corresponding test examples. labels are $0$ and $1$.



```python
from sklearn.datasets import make_classification, make_blobs
from sklearn.model_selection import train_test_split

# generate artificial data points
X, y = make_blobs(n_samples = 100,
                  n_features=2,
                  centers=[[5,5],[10,10]],
                  cluster_std=1.5,
                  random_state=2)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2,
                                                    random_state=123)
```

## Question 10

If $\hat{\mu}_0 = [\mu_1, \mu_2, ..., \mu_d]$ be the estimate for $\mu_0$, the mean of $0$ labeled examples, what will be the value of $\mu_1+\mu_2+...+\mu_d$? Write your answer correct to two decimal places.

We will be using the different covariances for different labeled examples. The estimate for $\Sigma_k$ will be

$$\hat{\Sigma}_k = \sigma_iI$$ where $\sigma_i$ is the variance of $i^{th}$ feature values of examples labeled $k$.



## Question 11
What will be value of $\text{trace}({\hat{\Sigma}}_0)$?  Write your answer correct to two decimal places.







```python
# Question 11
# Filter X_train for examples where y_train = 0 (already done in Question 10 as X_train_y0)
# X_train_y0 = X_train[y_train == 0]

# Calculate the variance for each feature for class 0
# The problem states Sigma_k = sigma_i*I where sigma_i is the variance of i-th feature
# The trace of this diagonal matrix will be the sum of the variances of each feature.
variances_y0 = np.var(X_train_y0, axis=0)

# Calculate the trace (sum of variances)
trace_sigma_0_hat = np.sum(variances_y0)

print(f"The value of trace(Sigma_0_hat) is: {trace_sigma_0_hat:.2f}")
```

## Question 12

Once we have estimated all the parameters for Gaussian naive Bayes assuming the different covariance matrices, we predict the labels for the training examples. What will be the training accuracy?

Accuracy is defined as the proportion of correctly classified examples.  Write your answer correct to two decimal places.




```python
# Enter your solution here
# Question 12

# Prior probabilities (from Question 9)
p_y1_gaussian = np.sum(y_train == 1) / len(y_train)
p_y0_gaussian = 1 - p_y1_gaussian

# Parameters for class 0 (from Question 10 and 11)
X_train_y0 = X_train[y_train == 0]
mu_0_hat = np.mean(X_train_y0, axis=0)
variances_y0 = np.var(X_train_y0, axis=0)

# Parameters for class 1
X_train_y1 = X_train[y_train == 1]
mu_1_hat = np.mean(X_train_y1, axis=0)
variances_y1 = np.var(X_train_y1, axis=0)

# Gaussian PDF function for a single feature
def gaussian_pdf(x, mean, variance):
    # Add a small epsilon to variance to avoid division by zero if variance is 0
    # This is a common practice in Naive Bayes to prevent log(0) or division by zero
    epsilon = 1e-9
    return (1 / np.sqrt(2 * np.pi * (variance + epsilon))) * np.exp(-((x - mean)**2) / (2 * (variance + epsilon)))

# Function to predict a single sample
def predict_gaussian_naive_bayes(sample, mu0, var0, mu1, var1, p0, p1):
    # Calculate likelihood for class 0
    likelihood_y0 = 1.0
    for i in range(len(sample)):
        likelihood_y0 *= gaussian_pdf(sample[i], mu0[i], var0[i])

    # Calculate likelihood for class 1
    likelihood_y1 = 1.0
    for i in range(len(sample)):
        likelihood_y1 *= gaussian_pdf(sample[i], mu1[i], var1[i])

    # Calculate posterior (proportional)
    posterior_y0 = likelihood_y0 * p0
    posterior_y1 = likelihood_y1 * p1

    return 0 if posterior_y0 > posterior_y1 else 1

# Make predictions on the training data
predictions_train = []
for i in range(len(X_train)):
    pred = predict_gaussian_naive_bayes(
        X_train[i], mu_0_hat, variances_y0, mu_1_hat, variances_y1,
        p_y0_gaussian, p_y1_gaussian
    )
    predictions_train.append(pred)

# Calculate training accuracy
predictions_train = np.array(predictions_train)
training_accuracy = np.sum(predictions_train == y_train) / len(y_train)

print(f"Training accuracy: {training_accuracy:.2f}")
```

## Question 13

What will be the test accuracy?

Accuracy is defined as the proportion of correctly classified examples.  




```python
# Question 13

# Make predictions on the test data
predictions_test = []
for i in range(len(X_test)):
    pred = predict_gaussian_naive_bayes(
        X_test[i], mu_0_hat, variances_y0, mu_1_hat, variances_y1,
        p_y0_gaussian, p_y1_gaussian
    )
    predictions_test.append(pred)

# Calculate test accuracy
predictions_test = np.array(predictions_test)
test_accuracy = np.sum(predictions_test == y_test) / len(y_test)

print(f"Test accuracy: {test_accuracy:.2f}")
```
